```python
import skimage as ski
import numpy as np
import napari
import glob
import scipy.ndimage as ndi
import tifffile
import cv2
import plotly.express as px
```

```python
viewer = napari.Viewer()
```

```python
root_dir = 'X:/arb/hp2655/20260818_PNX_IMARE-148212/'
```

```python
channels = ['ch1', 'ch2', 'ch3']
```

```python
channel = 0

fnames = glob.glob(root_dir + f'*/Images/*{channels[channel]}*.tiff')
```

```python
len(fnames)
```

```python
imgs = np.array([ski.io.imread(fname) for fname in fnames[::1000]])
```

```python
viewer.add_image(imgs)
```

```python
img_med = np.median(imgs, axis=0)
img_max = np.max(imgs, axis=0)
img_sum = np.sum(imgs, axis=0) / len(imgs)
```

```python
viewer.add_image(img_med)
viewer.add_image(img_max)
viewer.add_image(img_sum)
viewer.add_image(img_sum - img_med)
```

```python
for channel in range(len(channels)):
    print(channel)
    fnames = glob.glob(root_dir + f'*/Images/*{channels[channel]}*.tiff')
    imgs = np.array([ski.io.imread(fname) for fname in fnames[::1000]])
    img_med = np.median(imgs, axis=0)
    ski.io.imsave(channels[channel] + '_median.tif', img_med.astype(np.uint16))
```
